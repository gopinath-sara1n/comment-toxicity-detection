1. Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

2. Create the Model Performance Comparison Table

```python
comparison = pd.DataFrame({

    "Model":[
        "Logistic Regression",
        "BiLSTM",
        "GloVe + BiLSTM",
        "DistilBERT"
    ],

    "Hamming Loss":[
        0.028481,
        0.017262,
        0.016860,
        0.015032
    ],

    "Micro Precision":[
        0.570028,
        0.822722,
        0.794361,
        0.837504
    ],

    "Micro Recall":[
        0.864674,
        0.666233,
        0.720248,
        0.725014
    ],

    "Micro F1":[
        0.687095,
        0.736254,
        0.755492,
        0.777210
    ],

    "Macro F1":[
        0.562904,
        0.443658,
        0.587926,
        0.588496
    ],

    "ROC AUC":[
        0.980627,
        0.975242,
        0.983861,
        0.988290
    ]

})

comparison
```

3. Visualize Model Performance Comparison

```python
plt.figure(figsize=(8,5))

plt.bar(
    comparison["Model"],
    comparison["Micro F1"]
)

plt.title("Micro F1 Comparison")

plt.ylabel("Micro F1")

plt.xticks(rotation=20)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    comparison["Model"],
    comparison["Macro F1"]
)

plt.title("Macro F1 Comparison")

plt.ylabel("Macro F1")

plt.xticks(rotation=20)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    comparison["Model"],
    comparison["ROC AUC"]
)

plt.title("ROC-AUC Comparison")

plt.ylabel("ROC-AUC")

plt.xticks(rotation=20)

plt.show()
```

3. Create the Label-wise F1 Score Comparison Table

```python
label_f1 = pd.DataFrame({

    "Label":[
        "Toxic",
        "Severe Toxic",
        "Obscene",
        "Threat",
        "Insult",
        "Identity Hate"
    ],

    "BiLSTM":[
        0.79,
        0.37,
        0.81,
        0.00,
        0.69,
        0.00
    ],

    "GloVe + BiLSTM":[
        0.79,
        0.24,
        0.81,
        0.42,
        0.74,
        0.54
    ],

    "DistilBERT":[
        0.82,
        0.49,
        0.84,
        0.53,
        0.75,
        0.61
    ]

})

label_f1
```

4. Visualize Label-wise F1 Score Comparison

```python
label_f1.plot(
    x="Label",
    kind="bar",
    figsize=(10,5)
)

plt.title("Per-label F1 Score Comparison")

plt.ylabel("F1 Score")

plt.xticks(rotation=20)

plt.show()
```
